Мета: навчитися працювати з числовими типами даних та визначати тип змінних.


```python
a = 5
b = 2.5

c = a + b

print("Перше число:", a, "Тип:", type(a))
print("Друге число:", b, "Тип:", type(b))
print("Сума:", c, "Тип:", type(c))
```
